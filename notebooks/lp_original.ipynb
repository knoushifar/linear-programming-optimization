```python
from docplex.mp.model import Model
```

```python
lp = Model()
```

## 1.3

```python
x1 = lp.continuous_var()
x2 = lp.continuous_var() 
lp.maximize(x1 + 2*x2) 
lp.add_constraint(x1 - 4*x2 <= 4)
lp.add_constraint(-2*x1 + x2 <= 2)
lp.add_constraint(-3*x1 + 4*x2 <= 12)
lp.add_constraint(2*x1 + x2 <= 8)
sol = lp.solve() 
print(sol.objective_value) 
print(sol.get_value(x1)) 
print(sol.get_value(x2))
```

```python
sol = lp.solve()
if sol:
    print('z:', sol.objective_value) 
    print('x1:', sol.get_value(x1)) 
    print('x2:', sol.get_value(x2))
else:
    print('No solution')
```
